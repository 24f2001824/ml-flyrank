**Context:** `content_id` identifies the webpage. `client_id`, `provider_used`, and `model_used` provide contextual information about the data rather than the main content-performance signal.

**Excluded:** Identifier fields and categorical tier fields that duplicate information already represented by underlying measurements are excluded from the predictive feature set. ## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

The data does not contain a complete daily history for every webpage. Most of the performance data is aggregated into 90-day and 30-day windows.

It also does not show the result of an actual content refresh, so it cannot tell us whether refreshing a page caused an improvement in traffic, clicks or rankings.

The current and previous 30-day measurements are related periods for the same pages, so they should not be treated as independent observations.

The trend fields give directional information rather than a complete time series. The data is therefore useful for observed and directional analysis and for decision-support, but not for proving causal effects.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print(df[["days_with_impressions", "days_with_sessions"]].describe().T)

print("Pages with no impression days:",
      (df["days_with_impressions"] == 0).sum())

print("Pages with no session days:",
      (df["days_with_sessions"] == 0).sum())

display(df["trend_direction"].value_counts(dropna=False))
```
