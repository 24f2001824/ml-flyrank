## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### My lane as an ML task

This is a ranking task.

The goal is to rank webpages by their priority for content refresh. The model would produce a priority score for each webpage so that SEO specialists and content managers can review the highest-priority pages first.

Ranking is appropriate because the decision is not simply whether a page should be refreshed or not. The main decision is which pages should be reviewed first when the team has limited time and resources.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv(
    "/content/flyrank-ml-internship-starter/data/raw/content_refresh_anonymized.csv"
)

print("Dataset shape:", df.shape)
print("Unit of analysis: one webpage/content page")
print("Task type: Ranking")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

### Target or proxy

The target is a proxy for refresh priority rather than a directly observed refresh outcome.

The dataset contains measurable signals such as search impressions, clicks, CTR, average position, content age, days since the last update, and trend direction. These signals can be used to identify pages that may deserve review.

A higher priority score would indicate that a page appears more suitable for review based on these observed signals. This proxy should not be interpreted as proof that refreshing the page will improve traffic or rankings.

The dataset does not directly observe the causal outcome of refreshing a page, so the results will be treated as decision support rather than causal evidence.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

### Success metric

The primary success metric is Precision@50.

Precision@50 measures the proportion of the top 50 ranked webpages that are relevant according to the evaluation target.

A higher Precision@50 is better because the goal is to put the most useful refresh opportunities near the top of the ranking. For example, a Precision@50 of 0.70 would mean that approximately 70% of the top 50 recommendations are relevant according to the evaluation definition.

This metric matches the real decision because the content team has limited capacity and needs a useful shortlist rather than an accurate prediction for every webpage.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Why ML beats a fixed rule here

A fixed rule would have difficulty handling the many interacting signals in the dataset.

For example, a page may be old but its search trend may be improving, while another page may be relatively new but have declining impressions and a weak CTR. Search performance, freshness, engagement, and trend can therefore point in different directions.

An ML-based ranking approach can combine multiple signals and learn patterns from the available data instead of relying on a small number of manually chosen thresholds.

The goal is not to prove that ML causes better traffic or rankings, but to provide a more flexible ranking of pages for human review.
